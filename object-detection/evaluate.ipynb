# RUOD

```python
# REMEMBER CHANGING OUTPUT EVAL

from ultralytics import YOLO



model = YOLO("../runs/detect/outputs/training/RUOD/yolov26s_ruod_120e_fair/weights/best.pt")
results = model.val(
    data="../data-processing/vision/RUOD/ruod.yaml",
    split='test',
    imgsz=640,
    batch=1,
    iou=0.45, # nms
    conf= 0.25, # confidence threshold 

    plots=False,
    project="outputs/evaluation/RUOD",
    name="EVAL_yolov26s_ruod_120e_fair",
    exist_ok=True
)

```
